```python
import pandas as pd

dt = pd.read_csv("HR_Workforce_Analysis_Cleaned_Data.csv")
```

```python
dt.shape
```

```python
dt["Monthly_Income"] = dt["Monthly_Income"].str.replace("₹", "", regex=False).str.replace(",", "", regex=False).astype(float)

dt["Annual_Income"] = dt["Annual_Income"].str.replace("₹", "", regex=False).str.replace(",", "", regex=False).astype(float)
```

```python
dt.dtypes
```

```python
dt.describe()
```

```python
dt["Department"] = dt["Department"].str.title()

dt["Department"] = dt["Department"].replace("It", "IT")
```

```python
dt["Department"].value_counts()
```

```python
dt["Attrition"].value_counts()
```

```python
dt["Attrition"].value_counts(normalize=True) * 100
```

```python
pd.crosstab(dt["Department"], dt["Attrition"], normalize="index") * 100
```

```python
pd.crosstab(dt["Department"], dt["Attrition"])
```

```python
dt["Job_Role"].value_counts()
```

```python
pd.crosstab(dt["Job_Role"], dt["Attrition"], normalize="index") * 100
```

```python
pd.crosstab(dt["Job_Role"], dt["Attrition"])
```

```python
dt["Age_Group"] = pd.cut(
    dt["Age"],
    bins=[17, 25, 35, 45, 60],
    labels=["18-25", "26-35", "36-45", "46-60"]
)
```

```python
pd.crosstab(dt["Age_Group"], dt["Attrition"], normalize="index") * 100
```

```python
pd.crosstab(dt["Age_Group"], dt["Attrition"])
```

```python
pd.crosstab(dt["Job_Level"], dt["Attrition"], normalize="index") * 100
```

```python
pd.crosstab(dt["Job_Level"], dt["Attrition"])
```

```python
pd.crosstab(
    pd.cut(dt["Years_At_Company"], bins=[-1, 2, 5, 10, 20, 40],
           labels=["0-2", "3-5", "6-10", "11-20", "21+"]),
    dt["Attrition"],
    normalize="index"
) * 100
```
